```python
products = {
    "pen": {"price": 10, "stock": 20},
    "notebook": {"price": 50, "stock": 10},
    "bottle": {"price": 80, "stock": 8},
    "bag": {"price": 300, "stock": 5}
}

total_sales = 0

def show_stock():

    print("\n----- CURRENT STOCK -----")

    for item in products:

        price = products[item]["price"]
        stock = products[item]["stock"]

        print(item, "- Price:", price, "- Stock:", stock)


def sell_product():

    global total_sales

    item = input("Enter product name: ")
    item = item.lower()

    if item not in products:

        print("Item not found.")
        return

    try:

        quantity = int(input("Enter quantity: "))

    except ValueError:

        print("Please enter a numeric value.")
        return

    if quantity <= 0:

        print("Quantity must be greater than 0.")
        return

    if quantity > products[item]["stock"]:

        print("Not enough stock available.")
        print("Available stock:", products[item]["stock"])
        return

    price = products[item]["price"]

    sale_amount = price * quantity

    products[item]["stock"] = products[item]["stock"] - quantity

    total_sales = total_sales + sale_amount

    print("Sale successful!")
    print("Sale amount:", sale_amount)
    print("Remaining stock:", products[item]["stock"])


def save_data():

    with open("shop_data.txt", "w") as file:

        for item in products:

            price = products[item]["price"]
            stock = products[item]["stock"]

            file.write(item + "," + str(price) + "," + str(stock) + "\n")

        file.write("total_sales," + str(total_sales) + "\n")

    print("Data saved successfully.")


def load_data():

    global total_sales

    try:

        with open("shop_data.txt", "r") as file:

            lines = file.readlines()

        for line in lines:

            line = line.strip()

            parts = line.split(",")

            if parts[0] == "total_sales":

                total_sales = int(parts[1])

            else:

                item = parts[0]
                price = int(parts[1])
                stock = int(parts[2])

                products[item]["price"] = price
                products[item]["stock"] = stock

        print("Previous data loaded.")

    except FileNotFoundError:

        print("No previous data found. Starting with new stock.")


load_data()

while True:

    print("\n============================")
    print("      SMALL SHOP")
    print("============================")

    print("1. Show Stock")
    print("2. Record a Sale")
    print("3. Show Total Sales")
    print("4. Save Data")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        show_stock()

    elif choice == "2":

        sell_product()

    elif choice == "3":

        print("Total Sales:", total_sales)

    elif choice == "4":

        save_data()

    elif choice == "5":

        save_data()
        print("Thank you for using the program!")
        break

    else:

        print("Invalid choice. Please try again.")
```
